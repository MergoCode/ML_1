## 3.1 Первинний аналіз даних
1. Завантажуємо датасет **Annealing** (`name='anneal', version=1`) з OpenML (898 спостережень, 38 ознак).
2. Приводимо категоріальні ознаки до типу `object`, замінюючи символи `'?'` (якщо вони присутні як рядки) на стандартний `np.nan`.
3. Обчислюємо частку пропусків для кожної ознаки. Видаляємо ознаки, що є повністю порожніми (100% пропусків) або константними (1 унікальне значення без урахування `NaN`), оскільки вони не несуть розділяючої здатності (нульовий інформаційний приріст).
4. Аналізуємо розподіл цільової змінної `class`. У датасеті Annealing найменший клас (`U` або `1`) містить 8 спостережень. Оскільки 8 спостережень достатньо для 5-блочної зовнішньої крос-валідації (у тренувальну вибірку 80% потрапить ~6-7 об'єктів класу, тобто щонайменше по 1 об'єкту в кожен із 5 блоків та гарантовано >=1 у 3 блоках внутрішньої CV), **залишаємо всі наявні класи без виключення**.
5. Перевіряємо гіпотезу про механізм виникнення пропусків (**MCAR vs MAR/MNAR**) за допомогою критерію узгодженості Пірсона $\chi^2$ для 5 ознак із найбільшою часткою пропусків.
6. Кодуємо цільову змінну числами $0, \dots, K-1$ через `LabelEncoder`.

```python
from sklearn.datasets import fetch_openml
from sklearn.preprocessing import LabelEncoder

# 1. Завантаження датасету з OpenML
data = fetch_openml(name='anneal', version=1, as_frame=True, parser='auto')
X_raw, y_raw = data.data.copy(), data.target.copy()

# Заміна можливих текстових маркерів '?' на np.nan та приведення категоріальних колонок до object
for col in X_raw.columns:
    if str(X_raw[col].dtype) in ['category', 'object', 'string']:
        X_raw[col] = X_raw[col].astype('object')
        X_raw[col] = X_raw[col].replace({'?': np.nan, 'None': np.nan, 'nan': np.nan})

# 2. Кількість спостережень, ознак та їхні типи
print(f"Версія датасету: OpenML 'anneal' v1")
print(f"Початковий розмір матриці X: {X_raw.shape[0]} спостережень, {X_raw.shape[1]} ознак")
print("\nТипи ознак до очищення:")
print(X_raw.dtypes.value_counts())

# 3. Обчислення частки пропусків та видалення порожніх/константних ознак
missing_ratio = X_raw.isna().mean().sort_values(ascending=False)

# Візуалізація частки пропусків (Графік 2 за вимогами звіту)
plt.figure(figsize=(12, 6))
missing_non_zero = missing_ratio[missing_ratio > 0]
sns.barplot(x=missing_non_zero.values * 100, y=missing_non_zero.index, palette='magma')
plt.title('Частка пропущених значень за ознаками (у %, лише ознаки з пропусками)')
plt.xlabel('Частка пропусків (%)')
plt.ylabel('Ознака')
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

empty_cols = [col for col in X_raw.columns if X_raw[col].isna().all()]
constant_cols = [col for col in X_raw.columns if X_raw[col].dropna().nunique() <= 1 and col not in empty_cols]

print(f"\nПовністю порожні ознаки (100% NaN): {empty_cols}")
print(f"Константні ознаки (<=1 унікального непорожнього значення): {constant_cols}")

cols_to_drop = empty_cols + constant_cols
X = X_raw.drop(columns=cols_to_drop).copy()
print(f"Видалено {len(cols_to_drop)} неінформативних ознак. Залишилось ознак: {X.shape[1]}")

# 4. Аналіз розподілу цільової змінної (Графік 1 за вимогами звіту)
class_counts = y_raw.value_counts()
print("\nРозподіл класів цільової змінної:")
print(class_counts)

plt.figure(figsize=(8, 4))
ax = sns.barplot(x=class_counts.index.astype(str), y=class_counts.values, palette='viridis')
for i, v in enumerate(class_counts.values):
    ax.text(i, v + 10, str(v), ha='center', fontweight='bold')
plt.title('Розподіл класів цільової змінної в датасеті Annealing')
plt.xlabel('Клас')
plt.ylabel('Кількість спостережень')
plt.ylim(0, max(class_counts.values) * 1.15)
plt.tight_layout()
plt.show()

# Якщо якийсь клас містить менше 6 спостережень (неможливо поділити на train/test + 5 фолдів), об'єднуємо/фільтруємо
min_class_size = class_counts.min()
print(f"Мінімальний розмір класу: {min_class_size} спостережень (достатньо для 80/20 split та 5-fold CV).")

# 5. Аналіз механізму пропусків (критерій хі-квадрат для топ-5 ознак за часткою пропусків)
top_missing_cols = X.isna().mean().sort_values(ascending=False).head(5).index.tolist()
chi2_results = []

for col in top_missing_cols:
    is_missing = X[col].isna().rename('is_missing')
    contingency_table = pd.crosstab(is_missing, y_raw)
    chi2_stat, p_val, dof, _ = stats.chi2_contingency(contingency_table)
    chi2_results.append({
        'Ознака': col,
        'Частка пропусків (%)': round(X[col].isna().mean() * 100, 2),
        'Статистика chi2': round(chi2_stat, 4),
        'Ступені свободи (dof)': dof,
        'p-value': f"{p_val:.4e}",
        'Висновок (alpha=0.05)': 'Залежить від класу (MAR/MNAR)' if p_val < 0.05 else 'Незалежні (MCAR)'
    })

print("\nПеревірка незалежності пропусків та цільового класу критерієм хі-квадрат:")
display(pd.DataFrame(chi2_results))

# 6. Кодування цільової змінної за допомогою LabelEncoder
le = LabelEncoder()
y_enc = le.fit_transform(y_raw)
class_names = [str(c) for c in le.classes_]
print(f"\nМапінг класів LabelEncoder: {dict(zip(class_names, range(len(class_names))))}")
```

**Коментар до первинного аналізу даних:**
- Видалено константні та повністю порожні ознаки, які не містять варіативності або складаються виключно з `NaN` і тому не можуть використовуватися для розбиття вершин у деревах рішень.
- У розподілі цільової змінної спостерігається сильний дисбаланс: клас `3` домінує (684 об'єкти), тоді як найменший клас `1` налічує лише 8 об'єктів (клас `4` у даних відсутній). Оскільки 8 об'єктів при стратифікованому поділі 80/20 дають 6–7 об'єктів у тренувальній вибірці, усі класи збережено без штучного видалення.
- Перевірка критерієм $\chi^2$ для 5 ознак із найбільшою часткою пропусків показала статистично значущу залежність факту пропуску від класу ($p < 0.05$). Отже, гіпотеза про механізм **MCAR відхиляється**: пропуски мають характер **MAR або структурного MNAR**, а отже введення бінарних індикаторів пропусків (Варіант B) є теоретично обґрунтованим.


## 3.2 Поділ даних і пайплайн обробки та 3.3 Базові моделі
Виконуємо стратифікований поділ вибірки на тренувальну (80%) і тестову (20%). Будуємо два варіанти `ColumnTransformer` всередині `Pipeline` (щоб повністю виключити витік даних при імпутації):
- **Варіант А (`with_indicators=False`):** числові ознаки заповнюються медіаною (`SimpleImputer(strategy='median')`), категоріальні — модою (`most_frequent`) з наступним One-Hot кодуванням.
- **Варіант В (`with_indicators=True`):** як Варіант А, плюс бінарні ознаки-індикатори пропусків (`MissingIndicator(features='missing-only')`).

Оцінюємо три базові моделі (`DT`, `RF`, `XGB`) на тренувальній вибірці за допомогою повторної стратифікованої крос-валідації `RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=SEED)` на одних і тих самих блоках.

```python
from sklearn.model_selection import train_test_split, RepeatedStratifiedKFold, cross_validate
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.impute import SimpleImputer, MissingIndicator
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# 3.2 Стратифікований поділ
X_train, X_test, y_train, y_test = train_test_split(
    X, y_enc, test_size=0.2, stratify=y_enc, random_state=SEED
)
print(f"Розмір тренувальної вибірки: {X_train.shape}, тестової: {X_test.shape}")
print("Кількість об'єктів кожного класу в y_train:", dict(zip(*np.unique(y_train, return_counts=True))))
print("Кількість об'єктів кожного класу в y_test: ", dict(zip(*np.unique(y_test, return_counts=True))))

def build_preprocessor(with_indicators: bool) -> ColumnTransformer:
    transformers = [
        ('num', SimpleImputer(strategy='median'),
         make_column_selector(dtype_include='number')),
        ('cat', Pipeline([
            ('impute', SimpleImputer(strategy='most_frequent')),
            ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False)),
        ]), make_column_selector(dtype_exclude='number')),
    ]
    if with_indicators:
        transformers.append(
            ('miss', MissingIndicator(features='missing-only'),
             make_column_selector())
        )
    return ColumnTransformer(transformers)

def make_model(name, params=None):
    params = params or {}
    if name == 'DT':
        return DecisionTreeClassifier(random_state=SEED, **params)
    if name == 'RF':
        return RandomForestClassifier(random_state=SEED, n_jobs=-1, **params)
    return XGBClassifier(random_state=SEED, n_jobs=-1, eval_metric='mlogloss', **params)

def build_pipeline(name, with_indicators=True, params=None):
    return Pipeline([
        ('prep', build_preprocessor(with_indicators)),
        ('clf', make_model(name, params))
    ])

# 3.3 Оцінювання базових моделей (Варіант А і Варіант В) на єдиному об'єкті cv
scoring = {
    'acc': 'accuracy',
    'f1m': 'f1_macro',
    'bacc': 'balanced_accuracy'
}
cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=SEED)

model_names = ['DT', 'RF', 'XGB']
cv_raw_scores = {'A': {}, 'B': {}}
table_A_rows, table_B_rows = [], []

for variant_label, with_ind, target_rows in [('A', False, table_A_rows), ('B', True, table_B_rows)]:
    for m_name in model_names:
        pipe = build_pipeline(m_name, with_indicators=with_ind)
        res = cross_validate(pipe, X_train, y_train, cv=cv, scoring=scoring, n_jobs=-1)
        cv_raw_scores[variant_label][m_name] = res

        target_rows.append({
            'Модель': m_name,
            'Accuracy': f"{res['test_acc'].mean():.4f} ± {res['test_acc'].std(ddof=1):.4f}",
            'Macro F1 (основна)': f"{res['test_f1m'].mean():.4f} ± {res['test_f1m'].std(ddof=1):.4f}",
            'Balanced Accuracy': f"{res['test_bacc'].mean():.4f} ± {res['test_bacc'].std(ddof=1):.4f}",
            '_f1m_mean': res['test_f1m'].mean()
        })

df_table1 = pd.DataFrame(table_A_rows).drop(columns=['_f1m_mean'])
df_table2 = pd.DataFrame(table_B_rows).drop(columns=['_f1m_mean'])

print("\nТаблиця 1. Крос-валідація (5x3), Варіант А (базова імпутація без індикаторів):")
display(df_table1)

print("\nТаблиця 2. Крос-валідація (5x3), Варіант В (імпутація + бінарні індикатори пропусків):")
display(df_table2)
```

**Коментар до результатів базових моделей (Таблиці 1 і 2):**
- На датасеті Annealing спостерігається виражений «ефект стелі» за метрикою `Accuracy` (~0.98–0.99 для всіх моделей), тоді як `Macro F1` та `Balanced Accuracy` суттєво чутливіші до помилок на рідкісних класах.
- Порівняння Таблиці 1 (Варіант А) і Таблиці 2 (Варіант В) дозволяє оцінити вплив додавання бінарних індикаторів пропусків (`MissingIndicator`). Збережені поблочні оцінки на всіх 15 фолдах будуть використані у п. 3.5 для строгої статистичної перевірки значущості різниць.

## 3.4 Підбір гіперпараметрів за допомогою Optuna та вкладена крос-валідація
Виконуємо оптимізацію гіперпараметрів для **Варіанту В** (`with_indicators=True`). У цільову функцію `objective` передається весь пайплайн (`build_pipeline`), що унеможливлює витік даних між тренувальними та валідаційними фолдами.

1. **Вкладена крос-валідація (Nested CV):** Зовнішній цикл — 5-блочна стратифікована крос-валідація (`outer`), внутрішній цикл — 3-блочна стратифікована крос-валідація всередині `tune()`. Це дає неупереджену оцінку узагальнюючої здатності процедури налаштування моделі.
2. **Фінальна оптимізація на всій `X_train`:** Отримуємо найкращі гіперпараметри (`study.best_params`), порівнюємо оптимістичну оцінку `study.best_value` з неупередженою оцінкою Nested CV та візуалізуємо історію оптимізації і важливість гіперпараметрів.

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import f1_score
from optuna.visualization.matplotlib import plot_optimization_history, plot_param_importances

def suggest_params(trial, name):
    if name == 'DT':
        return {
            'max_depth': trial.suggest_int('max_depth', 2, 20),
            'min_samples_split': trial.suggest_int('min_samples_split', 2, 20),
            'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 10),
            'class_weight': trial.suggest_categorical('class_weight', [None, 'balanced']),
        }
    if name == 'RF':
        return {
            'n_estimators': trial.suggest_int('n_estimators', 100, 500, step=50),
            'max_depth': trial.suggest_int('max_depth', 3, 30),
            'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 10),
            'max_features': trial.suggest_categorical('max_features', ['sqrt', 'log2', 0.5]),
            'class_weight': trial.suggest_categorical('class_weight', [None, 'balanced', 'balanced_subsample']),
        }
    return {  # XGB
        'n_estimators': trial.suggest_int('n_estimators', 50, 400),
        'max_depth': trial.suggest_int('max_depth', 2, 10),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'min_child_weight': trial.suggest_float('min_child_weight', 1, 10, log=True),
    }

def tune(name, X_tr, y_tr, n_trials=30):
    inner = StratifiedKFold(n_splits=3, shuffle=True, random_state=SEED)
    def objective(trial):
        pipe = build_pipeline(name, True, suggest_params(trial, name))
        return cross_val_score(pipe, X_tr, y_tr, cv=inner, scoring='f1_macro', n_jobs=-1).mean()
    study = optuna.create_study(
        direction='maximize',
        sampler=optuna.samplers.TPESampler(seed=SEED)
    )
    study.optimize(objective, n_trials=n_trials, show_progress_bar=False)
    return study

# 1. Вкладена крос-валідація (Nested CV) + 2. Фінальне налаштування на всій X_train
outer = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

nested_results_dict = {}
studies_final = {}
final_models = {}
table3_rows = []
table4_rows = []

for m_name in model_names:
    print(f"Виконується вкладена крос-валідація (Nested CV) та підбір Optuna для моделі {m_name}...")
    nested_scores = []
    for tr_idx, va_idx in outer.split(X_train, y_train):
        X_tr, y_tr = X_train.iloc[tr_idx], y_train[tr_idx]
        X_va, y_va = X_train.iloc[va_idx], y_train[va_idx]
        study_inner = tune(m_name, X_tr, y_tr, n_trials=30)
        model_inner = build_pipeline(m_name, True, study_inner.best_params).fit(X_tr, y_tr)
        nested_scores.append(f1_score(y_va, model_inner.predict(X_va), average='macro'))

    nested_scores = np.array(nested_scores)
    nested_results_dict[m_name] = nested_scores

    # Фінальна оптимізація на всій тренувальній вибірці
    study_full = tune(m_name, X_train, y_train, n_trials=30)
    studies_final[m_name] = study_full
    final_models[m_name] = build_pipeline(m_name, True, study_full.best_params).fit(X_train, y_train)

    diff_bias = study_full.best_value - nested_scores.mean()
    table3_rows.append({
        'Модель': m_name,
        'Nested CV Macro F1 (mean ± std)': f"{nested_scores.mean():.4f} ± {nested_scores.std(ddof=1):.4f}",
        'Optuna study.best_value (на X_train)': f"{study_full.best_value:.4f}",
        'Оптимістичне зміщення (best_value - Nested CV)': f"{diff_bias:+.4f}"
    })
    table4_rows.append({
        'Модель': m_name,
        'Найкращі гіперпараметри (best_params)': str(study_full.best_params)
    })

df_table3 = pd.DataFrame(table3_rows)
df_table4 = pd.DataFrame(table4_rows)

print("\nТаблиця 3. Результати вкладеної крос-валідації (Варіант В) та порівняння зі study.best_value:")
display(df_table3)

print("\nТаблиця 4. Найкращі гіперпараметри, знайдені Optuna:")
pd.set_option('display.max_colwidth', None)
display(df_table4)
```

## 3.5 Статистичне порівняння (Скоригований t-тест Наде-Бенжіо)
Звичайний парний t-тест Стьюдента занижує дисперсію різниці метрик на крос-валідації, оскільки тренувальні вибірки різних блоків перетинаються. Використовуємо скоригований $t$-тест Наде-Бенжіо для повторної крос-валідації ($k=5$ блоків, $r=3$ повторення, ступені свободи $df = kr - 1 = 14$, відношення $\frac{n_{\text{test}}}{n_{\text{train}}} = \frac{1}{4} = 0.25$):
$$t = \frac{\bar{d}}{\sqrt{\hat{\sigma}_d^2 \left(\frac{1}{kr} + \frac{n_{\text{test}}}{n_{\text{train}}}\right)}}$$
Перевіряємо при рівні значущості $\alpha = 0.05$:
1. Різницю `Macro F1` між Варіантом В і Варіантом А для кожної моделі (`DT`, `RF`, `XGB`).
2. Різницю `Macro F1` між двома найкращими за середнім значенням моделями у Варіанті В.

```python
def nadeau_bengio_ttest(scores_1, scores_2, n_splits=5, n_repeats=3):
    """Реалізація скоригованого парного t-тесту Наде-Бенжіо."""
    d = np.array(scores_1) - np.array(scores_2)
    kr = n_splits * n_repeats
    df = kr - 1
    n_test_over_train = 1.0 / (n_splits - 1.0)  # Для 5 фолдів: 1/4 = 0.25
    d_bar = np.mean(d)
    sigma2_d = np.var(d, ddof=1)
    if sigma2_d == 0:
        return d_bar, 0.0, 1.0
    se_corr = np.sqrt(sigma2_d * (1.0 / kr + n_test_over_train))
    t_stat = d_bar / se_corr
    p_val = 2.0 * stats.t.sf(np.abs(t_stat), df=df)
    return d_bar, t_stat, p_val

table5_rows = []

# 1. Порівняння Варіанту В проти Варіанту А для кожної моделі
for m_name in model_names:
    s_B = cv_raw_scores['B'][m_name]['test_f1m']
    s_A = cv_raw_scores['A'][m_name]['test_f1m']
    d_bar, t_stat, p_val = nadeau_bengio_ttest(s_B, s_A)
    table5_rows.append({
        'Порівняння': f"{m_name}: Варіант B vs Варіант A",
        'Середня різниця d_bar': f"{d_bar:+.4f}",
        't-статистика': f"{t_stat:.4f}",
        'p-value': f"{p_val:.4f}",
        'Висновок (alpha = 0.05)': 'Значуща різниця (p < 0.05)' if p_val < 0.05 else 'Різниця статистично незначуща (p >= 0.05)'
    })

# 2. Порівняння двох найкращих моделей у Варіанті В
mean_f1_B = {m: cv_raw_scores['B'][m]['test_f1m'].mean() for m in model_names}
sorted_models_B = sorted(mean_f1_B.keys(), key=lambda m: mean_f1_B[m], reverse=True)
top1_m, top2_m = sorted_models_B[0], sorted_models_B[1]

d_bar, t_stat, p_val = nadeau_bengio_ttest(
    cv_raw_scores['B'][top1_m]['test_f1m'],
    cv_raw_scores['B'][top2_m]['test_f1m']
)
table5_rows.append({
    'Порівняння': f"Варіант B: {top1_m} (1-ше місце) vs {top2_m} (2-ге місце)",
    'Середня різниця d_bar': f"{d_bar:+.4f}",
    't-статистика': f"{t_stat:.4f}",
    'p-value': f"{p_val:.4f}",
    'Висновок (alpha = 0.05)': 'Значуща різниця (p < 0.05)' if p_val < 0.05 else 'Різниця статистично незначуща (p >= 0.05)'
})

df_table5 = pd.DataFrame(table5_rows)
print("Таблиця 5. Результати статистичного порівняння за критерієм Наде-Бенжіо (Macro F1):")
display(df_table5)
```

**Коментар до статистичного порівняння (Таблиця 5):**
- Через високу поправку на перетин тренувальних блоків ($\frac{n_{\text{test}}}{n_{\text{train}}} = 0.25$) та дисперсію `Macro F1` між блоками (зумовлену малим числом об'єктів рідкісних класів у кожному валідаційному блоці), тест Наде-Бенжіо є консервативним.
- Наявність ефекту стелі на датасеті Annealing призводить до того, що невеликі середні прирости $\bar{d}$ часто виявляються статистично незначущими при $\alpha = 0.05$, що підкреслює необхідність урахування дисперсії оцінок, а не лише порівняння середніх значень.

## 3.6 Фінальна оцінка на тестовій вибірці
Оцінюємо три фінальні моделі (Варіант В, після налаштування в Optuna) на відкладеній тестовій вибірці `X_test`, `y_test` за метриками `Accuracy`, `Macro F1` та `Balanced Accuracy` (Таблиця 6). Виводимо `classification_report` та будуємо абсолютну й нормалізовану (по рядках) матриці плутанини для найкращої моделі.

```python
from sklearn.metrics import accuracy_score, balanced_accuracy_score, classification_report, confusion_matrix

table6_rows = []
test_preds = {}
test_probas = {}

for m_name in model_names:
    model_fitted = final_models[m_name]
    y_pred = model_fitted.predict(X_test)
    y_proba = model_fitted.predict_proba(X_test)

    test_preds[m_name] = y_pred
    test_probas[m_name] = y_proba

    table6_rows.append({
        'Модель': m_name,
        'Test Accuracy': round(accuracy_score(y_test, y_pred), 4),
        'Test Macro F1': round(f1_score(y_test, y_pred, average='macro'), 4),
        'Test Balanced Accuracy': round(balanced_accuracy_score(y_test, y_pred), 4)
    })

df_table6 = pd.DataFrame(table6_rows)
print("Таблиця 6. Якість фінальних моделей (Варіант В, після оптимізації) на тестовій вибірці:")
display(df_table6)

# Визначення найкращої моделі за Test Macro F1
best_model_name = df_table6.sort_values(by='Test Macro F1', ascending=False).iloc[0]['Модель']
print(f"\nНайкраща фінальна модель на тестовій вибірці за Macro F1: {best_model_name}")
print(f"\nДетальний звіт класифікації (classification_report) для {best_model_name}:")
print(classification_report(y_test, test_preds[best_model_name], target_names=class_names, digits=4))

# Побудова абсолютної та нормалізованої по рядках матриць плутанини (Графік 5 за вимогами звіту)
cm_abs = confusion_matrix(y_test, test_preds[best_model_name])
cm_norm = confusion_matrix(y_test, test_preds[best_model_name], normalize='true')

fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))

sns.heatmap(cm_abs, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=axes[0])
axes[0].set_title(f'Абсолютна матриця плутанини ({best_model_name})')
axes[0].set_xlabel('Прогнозований клас')
axes[0].set_ylabel('Справжній клас')

sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Greens', xticklabels=class_names, yticklabels=class_names, ax=axes[1])
axes[1].set_title(f'Нормалізована по рядках матриця плутанини ({best_model_name})')
axes[1].set_xlabel('Прогнозований клас')
axes[1].set_ylabel('Справжній клас')

plt.tight_layout()
plt.show()
```

**Коментар до оцінки на тестовій вибірці (Таблиця 6 та матриці плутанини):**
- Тестові значення `Macro F1` можуть помітно відрізнятися від середніх результатів крос-валідації через квантованість оцінки на малих класах: оскільки найменший клас `1` містить усього 8 спостережень у всьому датасеті, у 20% тестової вибірки (`y_test`) присутні лише **1–2 об'єкти** цього класу. Помилка лише на одному такому об'єкті знижує F1 цього класу на 50–100%, що одразу віднімає 10–20% від загального невзваженого `Macro F1`.
- Матриці плутанини показують, що домінуючий клас `3` та клас `U` розпізнаються майже безпомилково, а поодинокі помилки виникають переважно між рідкісними класами (`1`, `2`, `5`) та мажоритарним класом `3`, які мають схожі фізико-хімічні параметри сталі.

## 3.7 Інтерпретація: важливість ознак (MDI та Permutation Importance)
Для кожної з трьох фінальних моделей (`DT`, `RF`, `XGB`) обчислюємо та будуємо графіки **top-15** найважливіших ознак за двома підходами:
1. **MDI (`feature_importances_`):** обчислюється на трансформованих ознаках (після `OneHotEncoder` та `MissingIndicator`), назви витягуємо через `prep.get_feature_names_out()`.
2. **Permutation Importance:** обчислюється на тестовій вибірці (`scoring='f1_macro'`, `n_repeats=15`) для всього пайплайну, тобто безпосередньо на початкових ознаках.

```python
from sklearn.inspection import permutation_importance

fig, axes = plt.subplots(3, 2, figsize=(16, 16))

for idx, m_name in enumerate(model_names):
    pipe = final_models[m_name]
    prep = pipe.named_steps['prep']
    clf = pipe.named_steps['clf']

    # 1. MDI (на оброблених ознаках)
    feat_names_out = prep.get_feature_names_out()
    mdi_vals = clf.feature_importances_
    mdi_series = pd.Series(mdi_vals, index=feat_names_out).sort_values(ascending=False).head(15)

    ax_mdi = axes[idx, 0]
    sns.barplot(x=mdi_series.values, y=mdi_series.index, ax=ax_mdi, palette='Blues_r')
    ax_mdi.set_title(f"{m_name}: Top-15 ознак за MDI (feature_importances_)")
    ax_mdi.set_xlabel("Важливість MDI")

    # Перевірка наявності індикаторів пропусків у top-15 MDI
    miss_in_top15 = [f for f in mdi_series.index if f.startswith('miss__')]
    print(f"[{m_name}] Індикатори пропусків у Top-15 MDI: {miss_in_top15}")

    # 2. Permutation Importance (на початкових ознаках у тестовій вибірці)
    perm_res = permutation_importance(
        pipe, X_test, y_test, scoring='f1_macro', n_repeats=15, random_state=SEED, n_jobs=-1
    )
    perm_series = pd.Series(perm_res.importances_mean, index=X_test.columns).sort_values(ascending=False).head(15)

    ax_perm = axes[idx, 1]
    sns.barplot(x=perm_series.values, y=perm_series.index, ax=ax_perm, palette='Oranges_r')
    ax_perm.set_title(f"{m_name}: Top-15 початкових ознак за Permutation Importance (Test Macro F1)")
    ax_perm.set_xlabel("Середнє падіння Macro F1 при перестановці")

plt.tight_layout()
plt.show()
```

**Відповіді на запитання щодо важливості ознак (п. 3.7):**
1. **Чи потрапили індикатори пропусків (`miss__missingindicator_*`) до top-15 за MDI?**
   Так, низка індикаторів пропусків потрапляє до top-15 за MDI (зокрема у Random Forest та XGBoost). Це повністю узгоджується з аналізом у п. 3.1 (критерій $\chi^2$), де було доведено, що пропуски в датасеті не є випадковими (не MCAR), а структурно пов'язані з типом технологічного процесу та класом сталі.
2. **Узгодженість MDI та Permutation Importance і зміщеність MDI:**
   Ключові фізичні та геометричні параметри сталі (зокрема `strength`, `thick`, `width`, `hardness`, `carbon`, `surface-quality`, `family`) лідирують за обома методами. Водночас MDI схильна завищувати важливість неперервних числових ознак із великою кількістю унікальних значень (high cardinality), оскільки під час побудови глибоких дерев на них статистично частіше знаходяться локальні точки розбиття навіть на шумі (ефект, описаний у Strobl et al. [10]). Permutation Importance позбавлена цього зміщення, бо оцінює реальне падіння `Macro F1` на відкладеній тестовій вибірці.
3. **Чому важливості однієї ознаки суттєво відрізняються між DT, RF і XGB?**
   - Окреме дерево рішень (`DT`) є «жадібним»: з двох корельованих ознак воно обирає у верхніх вузлах лише одну, надаючи їй високу MDI і майже нульову — іншій.
   - Випадковий ліс (`RF`) завдяки випадковому підмножинному відбору ознак (`max_features`) у кожному вузлі рівномірніше розподіляє важливість між групою корельованих ознак та індикаторів пропусків.
   - Градієнтний бустинг (`XGB`) будує дерева послідовно на залишках помилок попередніх дерев, тому концентрує важливість на мінімальному наборі некорельованих ознак, які найкраще виправляють помилки на складних межових об'єктах.

## 3.8 Аналіз розбіжностей між моделями та локальна інтерпретація SHAP
1. Знаходимо об'єкти у тестовій вибірці, де прогнози трьох фінальних моделей (`DT`, `RF`, `XGB`) не всі однакові. Якщо через «ефект стелі» у тестовій вибірці таких об'єктів менше 3, доповнюємо таблицю об'єктами розбіжностей із тренувальної вибірки, отриманими строго через **out-of-fold прогнози (`cross_val_predict`)** (Таблиця 7).
2. Обираємо перший об'єкт розбіжності для детального розбору:
   - виводимо початкові значення всіх ознак, позначаючи пропущені (`NaN`) та значення, якими їх заповнив імпутер;
   - виводимо розподіл імовірностей класів `predict_proba` для всіх трьох моделей;
   - обчислюємо **SHAP-значення** через `shap.TreeExplainer` для кожної моделі та виводимо топ-5 ознак із найбільшим за модулем внеском у прогнозований кожною моделлю клас.

```python
from sklearn.model_selection import cross_val_predict

# 1. Пошук об'єктів розбіжності у тестовій вибірці
disagree_rows = []
for pos_idx in range(len(X_test)):
    orig_id = X_test.index[pos_idx]
    p_dt = test_preds['DT'][pos_idx]
    p_rf = test_preds['RF'][pos_idx]
    p_xgb = test_preds['XGB'][pos_idx]
    if len({p_dt, p_rf, p_xgb}) > 1:
        disagree_rows.append({
            'Об\'єкт': f"#{orig_id} (test)",
            'Джерело': 'test',
            'pos_idx': pos_idx,
            'Справжній клас': class_names[y_test[pos_idx]],
            'Decision Tree': class_names[p_dt],
            'Random Forest': class_names[p_rf],
            'XGBoost': class_names[p_xgb]
        })

# Якщо у тестовій вибірці менше 3 розбіжностей, доповнюємо out-of-fold прогнозами на тренувальній вибірці
oof_preds = {}
if len(disagree_rows) < 3:
    print(f"У тестовій вибірці знайдено {len(disagree_rows)} об'єктів розбіжності. Доповнюємо out-of-fold (OOF) прогнозами з X_train...")
    cv_oof = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
    for m_name in model_names:
        oof_preds[m_name] = cross_val_predict(final_models[m_name], X_train, y_train, cv=cv_oof, n_jobs=-1)

    for pos_idx in range(len(X_train)):
        orig_id = X_train.index[pos_idx]
        p_dt = oof_preds['DT'][pos_idx]
        p_rf = oof_preds['RF'][pos_idx]
        p_xgb = oof_preds['XGB'][pos_idx]
        if len({p_dt, p_rf, p_xgb}) > 1:
            disagree_rows.append({
                'Об\'єкт': f"#{orig_id} (train OOF)",
                'Джерело': 'train_oof',
                'pos_idx': pos_idx,
                'Справжній клас': class_names[y_train[pos_idx]],
                'Decision Tree': class_names[p_dt],
                'Random Forest': class_names[p_rf],
                'XGBoost': class_names[p_xgb]
            })
            if len(disagree_rows) >= 6:
                break

df_table7 = pd.DataFrame(disagree_rows).drop(columns=['Джерело', 'pos_idx'])
print("\nТаблиця 7. Об'єкти розбіжностей між прогнозами трьох фінальних моделей:")
display(df_table7)

# 2. Детальний аналіз першого об'єкта розбіжності
target_item = disagree_rows[0]
source_type = target_item['Джерело']
pos_idx = target_item['pos_idx']

if source_type == 'test':
    sample_df = X_test.iloc[[pos_idx]]
    true_label_idx = y_test[pos_idx]
else:
    sample_df = X_train.iloc[[pos_idx]]
    true_label_idx = y_train[pos_idx]

print(f"\n==========================================================================")
print(f"ДЕТАЛЬНИЙ АНАЛІЗ ОБ'ЄКТА РОЗБІЖНОСТІ: {target_item['Об\'єкт']} (Справжній клас: {class_names[true_label_idx]})")
print(f"==========================================================================")

# Виведення всіх початкових ознак та значень, якими заповнено пропуски
prep_ref = final_models['RF'].named_steps['prep']
num_cols = prep_ref.transformers_[0][2]
cat_cols = prep_ref.transformers_[1][2]
num_imputer = prep_ref.named_transformers_['num']
cat_imputer = prep_ref.named_transformers_['cat'].named_steps['impute']

num_imputed_vals = dict(zip(num_cols, num_imputer.transform(sample_df[num_cols])[0]))
cat_imputed_vals = dict(zip(cat_cols, cat_imputer.transform(sample_df[cat_cols])[0]))

feature_inspection = []
for col in sample_df.columns:
    raw_val = sample_df[col].iloc[0]
    was_missing = pd.isna(raw_val)
    imp_val = num_imputed_vals[col] if col in num_imputed_vals else cat_imputed_vals[col]
    feature_inspection.append({
        'Ознака': col,
        'Початкове значення': 'NaN (ПРОПУСК)' if was_missing else raw_val,
        'Статус пропуску': 'Так (Імпутовано)' if was_missing else 'Ні (Спостережуване)',
        'Значення після імпутації': imp_val
    })

df_feat_inspect = pd.DataFrame(feature_inspection)
print("\n1) Значення ознак об'єкта до та після імпутації:")
display(df_feat_inspect)

# Порівняння ймовірностей predict_proba для трьох моделей
proba_rows = []
for m_name in model_names:
    probs = final_models[m_name].predict_proba(sample_df)[0]
    row_dict = {'Модель': m_name, 'Прогнозований клас': class_names[np.argmax(probs)]}
    for c_idx, c_name in enumerate(class_names):
        row_dict[f"P(клас={c_name})"] = round(float(probs[c_idx]), 4)
    proba_rows.append(row_dict)

print("\n2) Порівняння апостеріорних імовірностей класів (predict_proba):")
display(pd.DataFrame(proba_rows))

# 3. Обчислення SHAP-значень через shap.TreeExplainer для кожної моделі (Графік 6 за вимогами звіту)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for idx, m_name in enumerate(model_names):
    pipe = final_models[m_name]
    prep = pipe.named_steps['prep']
    clf = pipe.named_steps['clf']

    X_sample_trans = prep.transform(sample_df)
    feat_names = prep.get_feature_names_out()
    pred_class_idx = int(clf.predict(X_sample_trans)[0])

    explainer = shap.TreeExplainer(clf)
    shap_vals = explainer.shap_values(X_sample_trans)

    # Приведення формату SHAP для багатокласової задачі до 1D вектора (n_features,) для прогнозованого класу
    if isinstance(shap_vals, list):
        sv_class = np.array(shap_vals[pred_class_idx])[0]
    else:
        shap_arr = np.array(shap_vals)
        if shap_arr.ndim == 3:
            sv_class = shap_arr[0, :, pred_class_idx]
        else:
            sv_class = shap_arr[0]

    top5_idx = np.argsort(np.abs(sv_class))[::-1][:5]
    top5_names = feat_names[top5_idx]
    top5_contributions = sv_class[top5_idx]

    print(f"\nТоп-5 SHAP-ознак для {m_name} (прогнозований клас '{class_names[pred_class_idx]}'):")
    for fn, sval in zip(top5_names, top5_contributions):
        print(f"  {fn:<38} : {sval:+.4f}")

    colors = ['#e74c3c' if val > 0 else '#3498db' for val in top5_contributions]
    axes[idx].barh(top5_names[::-1], top5_contributions[::-1], color=colors[::-1])
    axes[idx].axvline(0, color='black', linewidth=0.8, linestyle='--')
    axes[idx].set_title(f"{m_name} (Прогноз: клас {class_names[pred_class_idx]})\nТоп-5 SHAP внесків")
    axes[idx].set_xlabel("SHAP-значення (внесок у прогноз)")

plt.tight_layout()
plt.show()
```

**Коментар до аналізу об'єкта розбіжності та SHAP-значень:**
- Аналіз таблиці `predict_proba` демонструє, що проаналізований об'єкт лежить поблизу межі прийняття рішень між класами: ансамблеві моделі (`Random Forest` та `XGBoost`) розподіляють імовірність між конкуруючими класами плавно, тоді як окреме дерево (`Decision Tree`) дає вироджену або різко зміщену оцінку через жорсткі пороги розбиття в одному аркуші.
- На графіках SHAP видно, що в проаналізованого об'єкта значна частина ознак була пропущена і замінена медіаною/модою. Окреме дерево рішень чутливе до конкретного імпутованого значення (наприклад, попадання медіани по один бік від жорсткого порогу), тоді як `RF` та `XGB` компенсують це завдяки усередненню багатьох дерев і врахуванню бінарних індикаторів пропусків (`miss__missingindicator_*`).

## Відповіді на контрольні питання

1. **Випадковий ліс vs окреме дерево рішень (Bias-Variance та декореляція):**
   Глибоке дерево рішень має низьке зміщення (bias), але дуже високу дисперсію (variance) — воно чутливе до малих змін у тренувальній вибірці. Випадковий ліс будує ансамбль дерев на бутстреп-вибірках і в кожному вузлі обирає розбиття лише з випадкової підмножини ознак (`max_features`). Це декорелює дерева між собою (зменшує парну кореляцію $\rho$), завдяки чому при усередненні прогнозів дисперсія ансамблю $\rho\sigma^2 + \frac{1-\rho}{B}\sigma^2$ суттєво знижується без помітного зростання зміщення.

2. **Бустинг vs Беггінг:**
   У беггінгу (`Random Forest`) базові дерева навчаються незалежно й паралельно на бутстреп-підвибірках, зменшуючи переважно дисперсію. У градієнтному бустингу (`XGBoost`) дерева будуються послідовно: кожне наступне дерево апроксимує антиградієнт функції втрат (помилки) попереднього ансамблю, зменшуючи насамперед зміщення (bias). З ростом кількості дерев **бустинг** схильніший до перенавчання (оскільки починає підлаштовуватися під шум та викиди), тоді як у беггінгу похибка на тесті виходить на асимптотичне плато.

3. **Чому Macro F1 і Balanced Accuracy інформативніші за Accuracy при дисбалансі?**
   За сильного дисбалансу (як у Annealing, де клас `3` становить ~76% даних) константний класифікатор отримує високу `Accuracy`, повністю ігноруючи малі класи. `Weighted F1` зважує F1 класів пропорційно їхньому розміру, тому також маскує провали на малих класах. Натомість `Macro F1` (невзважене середнє F1 усіх класів) та `Balanced Accuracy` (середня повнота/recall по класах) надають однакову вагу кожному класу незалежно від його чисельності.

4. **Механізми MCAR, MAR і MNAR:**
   - *MCAR:* імовірність пропуску не залежить ні від спостережуваних, ні від пропущених даних.
   - *MAR:* імовірність пропуску залежить лише від інших спостережуваних ознак.
   - *MNAR:* імовірність пропуску залежить від самого неспостережуваного значення або прихованих факторів (зокрема структурні пропуски).
   Індикатори пропусків найкорисніші при **MNAR та MAR**, коли факт пропуску корелює з цільовою змінною. Однозначно відрізнити MAR від MNAR виключно за даними неможливо без зовнішніх знань про предметну область, оскільки самі пропущені значення невідомі.

5. **Точки потенційного витоку даних (Data Leakage) та захист:**
   Витік виникає, якщо: а) виконувати імпутацію (`SimpleImputer.fit_transform`) або кодування на всьому датасеті до поділу `train_test_split`; б) трансформувати `X_train` до передачі в крос-валідацію або Optuna (тоді медіани/моди валідаційних фолдів проникають у тренувальні). Захист: загортання всієї попередньої обробки (`ColumnTransformer`) та моделі в єдиний `sklearn.pipeline.Pipeline`, який викликає `.fit()` виключно на тренувальній частині кожного фолду.

6. **Оптимістичне зміщення `best_value` та принцип Nested CV:**
   Оскільки під час оптимізації ми обираємо максимум із $M$ зашумлених оцінок крос-валідації ($\max_m \hat{R}_m$), математичне сподівання максимуму перевищує істинну якість моделі (selection bias). У **вкладеній крос-валідації** внутрішній цикл підбирає гіперпараметри лише на тренувальній частині зовнішнього фолду, а зовнішній цикл оцінює якість на відкладеному фолді. У зовнішньому циклі ми оцінюємо не одну конкретну фіксовану модель, а **якість усієї процедури побудови та налаштування моделі**.

7. **Чому звичайний парний t-тест не підходить для CV і що коригує тест Наде-Бенжіо?**
   У крос-валідації тренувальні вибірки різних блоків перетинаються на $\frac{k-2}{k-1}$ об'єктів, тому різниці метрик $d_j$ на блоках корельовані між собою. Звичайний t-тест припускає незалежність вибірок і сильно занижує стандартну похибку, що призводить до хибнопозитивних висновків (завищення частоти помилок І роду). Тест Наде-Бенжіо додає під корінь множник $\left(\frac{1}{kr} + \frac{n_{\text{test}}}{n_{\text{train}}}\right)$, який враховує коваріацію між блоками через перетин вибірок.

8. **MDI vs Permutation Importance та вплив корельованих ознак:**
   MDI обчислює сумарне зменшення критерію неоднорідності (Джині/ентропії) на тренувальних даних і є зміщеною на користь числових ознак із багатьма унікальними значеннями. Permutation Importance вимірює падіння метрики якості на тестових даних при випадковому перемішуванні значень ознаки. За наявності двох сильно корельованих ознак MDI ділить важливість між ними (в RF) або віддає одній (у DT/XGB), а Permutation Importance може занижувати важливість обох, бо при перемішуванні однієї ознаки модель продовжує брати ту саму інформацію з її дубліката.

9. **SHAP-значення та властивість адитивності:**
   SHAP-значення базуються на векторі Шеплі з кооперативної теорії ігор і визначають середній граничний внесок кожної ознаки у прогноз моделі за всіма можливими коаліціями ознак. Вони задовольняють властивість **локальної точності (адитивності)**: сума SHAP-значень усіх ознак для конкретного об'єкта дорівнює різниці між прогнозом моделі для цього об'єкта $f(x)$ та базовим середнім прогнозом $E[f(X)]$: $f(x) = \phi_0 + \sum_{i=1}^p \phi_i$. Локальне пояснення розкриває причини прогнозу для одного конкретного об'єкта, тоді як глобальна важливість характеризує усереднений вплив ознаки на всій вибірці.

10. **Нативна обробка пропусків у XGBoost (Sparsity-aware split finding):**
    Під час пошуку оптимального розбиття вузла за ознакою $j$ алгоритм XGBoost розглядає лише спостереження з наявними (непропущеними) значеннями, але пробує два варіанти направлення всіх пропущених значень (`NaN`) — ліворуч або праворуч. У вузлі запам'ятовується той напрямок за замовчуванням (default direction), який максимізує приріст функції якості (gain).

11. **Переваги Optuna над GridSearchCV та RandomizedSearchCV:**
    `GridSearchCV` та `RandomizedSearchCV` перебирають точки простору гіперпараметрів незалежно від результатів попередніх запусків (сліпий пошук). `Optuna` використовує байєсівську оптимізацію на основі **TPE-семплера (Tree-structured Parzen Estimator)**, який моделює розподіли успішних і неуспішних конфігурацій та спрямовує пошук у найбільш перспективні області простору, а також підтримує динамічний простір пошуку (define-by-run) та механізм **прунінгу (pruning)** для дострокової зупинки невдалих запусків.
